```python
# Import necessary libraries
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Load the dataset
file_path = 'heart_statlog_cleveland_hungary_final.csv'
df = pd.read_csv(file_path)

# Display the first few rows of the dataset to understand its structure
df.head()
```

```python
# Set the figure size
plt.figure(figsize=(16, 5))

# Plot histograms for Age, Cholesterol, and Maximum Heart Rate (assuming column names are 'age', 'chol', 'thalach')
plt.subplot(1, 3, 1)
plt.hist(df['age'], bins=20, color='skyblue')
plt.title('Age Distribution')
plt.xlabel('Age')
plt.ylabel('Frequency')

plt.subplot(1, 3, 2)
plt.hist(df['cholesterol'], bins=20, color='lightgreen')
plt.title('Cholesterol Levels Distribution')
plt.xlabel('Cholesterol')
plt.ylabel('Frequency')

plt.subplot(1, 3, 3)
plt.hist(df['max heart rate'], bins=20, color='lightcoral')
plt.title('Max Heart Rate Distribution')
plt.xlabel('Max Heart Rate')
plt.ylabel('Frequency')

plt.tight_layout()
plt.show()
```

```python
# Plot the class distribution (heart disease cases)
plt.figure(figsize=(6, 6))
sns.countplot(x='target', data=df, color='lightblue')
plt.title('Distribution of Heart Disease Cases')
plt.xlabel('Heart Disease (1 = Disease, 0 = No Disease)')
plt.ylabel('Count')
plt.show()
```

```python
# Display statistical summary (mean, standard deviation, etc.)
stat_summary = df[['age', 'cholesterol', 'max heart rate']].describe()
print(stat_summary)

# Correlation matrix
correlation_matrix = df[['age', 'cholesterol', 'max heart rate']].corr()
print("Correlation Matrix:\n", correlation_matrix)

# Visualize the correlation matrix using a heatmap
plt.figure(figsize=(8, 6))
sns.heatmap(correlation_matrix, annot=True, cmap='coolwarm', fmt=".2f")
plt.title('Correlation Matrix')
plt.show()
```
